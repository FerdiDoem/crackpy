# Williams and CJP Method-Fit Usage Demo

This notebook demonstrates the deterministic method-module APIs for the Williams fit, the CJP mixed-mode fit, and the CJP Mode-I fit.

It covers:

- typed result assembly from explicit coefficient arrays;
- fitting already-sampled crack-tip-centered displacement fields;
- firing the legacy optimizer-adapter entry points;
- building result/provenance envelopes;
- writing envelope, compact KG, graph JSON, and HTML graph artifacts.

Run from the repository root or keep the path bootstrap cell below.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
project_root = cwd if (cwd / "crackpy").exists() else cwd.parents[1]
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

project_root
```

```python
from scripts.fracture_analysis.method_fit_usage_demo import (
    DEFAULT_OUTPUT_DIR,
    demonstrate_cjp_fit,
    demonstrate_williams_fit,
    run_demo,
)

output_dir = DEFAULT_OUTPUT_DIR
output_dir
```

## Williams Fit

`demonstrate_williams_fit()` shows three ways to fire the Williams method module:

1. `run_williams_fit_from_coefficients(...)` for explicit fitted coefficient arrays;
2. `fit_williams_displacement_field(...)` for already-sampled crack-tip-centered displacement fields;
3. `run_williams_fit(...)` for a legacy optimizer-compatible object.

It then builds a Williams result/provenance envelope and writes the artifact projections.

```python
williams = demonstrate_williams_fit(output_dir)
williams_result = williams["result_from_field"]

{
    "K_I": williams_result.K_I,
    "K_II": williams_result.K_II,
    "T": williams_result.T,
    "error_xy": williams_result.error_xy,
    "artifacts": {name: str(path.relative_to(project_root)) for name, path in williams["artifacts"].items()},
}
```

## CJP Mixed-Mode and CJP Mode-I Fits

`demonstrate_cjp_fit()` shows both CJP method variants:

- `cjp_mixed_mode_outputs_from_coefficients(...)` for mixed-mode coefficients `(A_r, B_r, B_i, C, E)`;
- `cjp_mode_i_outputs_from_coefficients(...)` for Mode-I coefficients `(A, B, C, E, F)`;
- `run_cjp_fit_from_coefficients(...)` to assemble both variants into one typed CJP result;
- `fit_cjp_displacement_field(...)` to fit both variants from explicit displacement arrays;
- `run_cjp_fit(...)` to fire the legacy optimizer-compatible adapter.

It then builds one CJP envelope with two method records, two runs, and two result records.

```python
cjp = demonstrate_cjp_fit(output_dir)
cjp_result = cjp["result_from_field"]

{
    "mixed_mode": {
        "K_F": cjp_result.mixed_mode.K_F,
        "K_R": cjp_result.mixed_mode.K_R,
        "K_S": cjp_result.mixed_mode.K_S,
        "K_II": cjp_result.mixed_mode.K_II,
        "T": cjp_result.mixed_mode.T,
        "error": cjp_result.mixed_mode.error,
    },
    "mode_i": {
        "K_F": cjp_result.mode_i.K_F,
        "K_R": cjp_result.mode_i.K_R,
        "K_S": cjp_result.mode_i.K_S,
        "T_x": cjp_result.mode_i.T_x,
        "T_y": cjp_result.mode_i.T_y,
        "error": cjp_result.mode_i.error,
    },
    "artifacts": {name: str(path.relative_to(project_root)) for name, path in cjp["artifacts"].items()},
}
```
